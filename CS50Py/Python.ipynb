```python
# Typecasting
x= float(input("Enter first number"))
y= float(input("Enter second number"))
z = x+y

print(z)
# Round function syntax - round(number, digits to be rounded off to )
print(round(z))
print(round(z,3))
print(f"{z:.2f}") # rounding off to 2 digits using f strings instead of round function

print(f"{z:,}") # to write with comma after every 3 digits like US notation
```

```python
def greet(person):
    match person:
        case "A" | "C":
            print("Hello, buddy!")
        case "B":
            print("Hello, sir!")
        case _:
            print("Hello, stranger!")

greet("A")
greet("B")
greet("C")
greet("Rishit")
```

# Try Except block

```python
def get_int():
    # function takes integer input from the user and reprompts if they enter something else. (uses try except)

    while True:
        try:
            a = int(input("Enter a Number"))
        except ValueError:
            print("Please Enter an integer")
        else:
            return a

print(get_int())
```
